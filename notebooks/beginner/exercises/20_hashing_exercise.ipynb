# 1. Hashing Algorithms 
A hash is a result generated by applying an algorithm or numeric process to a value.
They are often used to store a password in a file as a hash string. Hash functions are one way and can’t be reversed so if hacker gains access to database they cannot access user’s password.
Good hash algorithm should have low chance of collision (generating same output from different inputs) and be quick to calculate. Writing complex hash algorithms is a quite a lucrative business!

## Using Python to Generate a Hash
The following code will generate the md5 hash result on a string.

```python
import hashlib
msg="hello"
result = hashlib.md5(msg.encode()).hexdigest()
print(result)
```
