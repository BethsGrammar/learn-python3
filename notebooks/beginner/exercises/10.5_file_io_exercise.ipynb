# requests
Another way to get data into your python script is to use pull data from the web.
In future lessons, we will look at API requests. In this exercise we will use requests to open a file stored in this github repo into our Python script.

To do this we need to 
- `import requests`
- read the data using the `get(url)` method. This will collect a response from the given url argument 
- prep the data for our script using `.text.splitlines()`
    - `.text` gathers the data from the response.
    - `.splitlines()` splits the data into line where there are `'\n'` in the data. 

See the example below:

```python
import requests
url = 'https://raw.githubusercontent.com/BethsGrammar/learn-python3/refs/heads/master/notebooks/beginner/exercises/assets/oasis_songs.txt'
file = requests.get(url).text.splitlines()

print(file)
```
